```python
import pandas as pd
```

```python
def check_detection_identification(csv_ori_path, csv_result_path, dict_pred_true, dict_true_map):
    # merge csv
    df_ori = pd.read_csv(csv_ori_path)
    df_result = pd.read_csv(csv_result_path)
    df_ori["true_brand"] = df_ori["brand"]
    df = pd.merge(df_result, df_ori[["true_brand", "scr_path"]], on="scr_path", how="inner")  # 'inner' keeps only matching rows
    df["pred_brand"] = df["only_name"].apply(lambda x: x.split("/")[-2])
    # 0 as benign, 1 as phish
    # detection
    df_phish = df[df["phish"] == 1]
    print(f"total: {len(df)}, pred as phish: {len(df_phish)} {len(df_phish)/len(df)}, pred as benign: {len(df) - len(df_phish)} {(len(df) - len(df_phish))/len(df)}")
    # identification
    df_phish['map_pred_brand'] = df_phish['pred_brand'].copy().map(dict_pred_true).fillna(df_phish['pred_brand'])
    df_phish['map_true_brand'] = df_phish['true_brand'].copy().map(dict_true_map).fillna(df_phish['true_brand'])

    df_ident = df_phish[df_phish["map_true_brand"]==df_phish["map_pred_brand"]]
    print(f"Itp: {len(df_ident)}, Itp/phish: {len(df_ident)/len(df_phish)}")
    
    # check the non identified results
    key_column = "scr_path"  

    df_non_ident = df_phish[~df_phish[key_column].isin(df_ident[key_column])]

    # df_non_ident[['true_brand','pred_brand','phish','tagBox','siamese_conf','url','map_pred_brand','map_true_brand']].to_csv("aaa.csv") # [['map_true_brand','map_pred_brand']]
```
